### Analisis de la red tipo T

<img src="./topologia_T.png" />

Para analizar esta red nos conviene empezar por los parametros de impedancia donde conectamos una fuente de corriente en la salida y otra en la entrada, de forma tal que la corriente "entre" al nodo común de las impedancias y usando la defifinicion de los parametro de impedancia procedemos a calcular nuestramatriz de impedancia.

$$V_1 = I_1 \cdot Z_{11} + I_2 \cdot Z_{12} $$
$$V_2 = I_1 \cdot Z_{21} + I_2 \cdot Z_{22} $$

Donde:

$$
\begin{aligned}
Z_{11} = \left. \frac{V_1}{I_1} \right|_{I_2 = 0} &\hspace{5em} Z_{12} = \left. \frac{V_1}{I_2} \right|_{I_1 = 0} \\
\\
Z_{21} = \left. \frac{V_2}{I_1} \right|_{I_2 = 0} &\hspace{5em} Z_{22} = \left. \frac{V_2}{I_2} \right|_{I_1 = 0} \\
\end{aligned}
$$

Por lo que las componentes de las matrices que de la siguiente forma.

$$
\begin{aligned}
Z_{11} = \left. \frac{V_1}{I_1} \right|_{I_2 = 0} = Z_A + Z_B &\hspace{5em} Z_{12} = \left. \frac{V_1}{I_2} \right|_{I_1 = 0} = Z_B\\
\\
Z_{22} = \left. \frac{V_2}{I_2} \right|_{I_1 = 0} = Z_B + Z_C &\hspace{5em} Z_{21} = \left. \frac{V_2}{I_1} \right|_{I_2 = 0} = Z_B\\
\end{aligned}
$$

Quedando nuestra matriz de parametros Zdela siguiente forma:

$$
Z =
\begin{bmatrix}
Z_{11}   &    Z_{12}  \\
Z_{21}   &   Z_{22}     
\end{bmatrix} = 
\begin{bmatrix}
Z_A + Z_B   &    Z_B  \\
Z_B   &   Z_B + Z_C     
\end{bmatrix}
$$

Con nuestra matriz Z podemos calcular nuestra matriz de admitancias Y, donde sabemos que $Y = Z^-1$ y al ser Z una matriz tenemos que su inversa se representa de la siguiente forma:

$$
Y = \frac{1}{det(Z)} \cdot Adj(A)
$$

* Calculo de **det(Z)**

$$
det(Z)= (Z_A+Z_B) \cdot (Z_C+Z_B) - Z_B^2
$$

$$
\require{cancel}
det(Z)= Z_A \cdot Z_C + Z_A \cdot Z_B + Z_B \cdot Z_C + \cancel{Z_B^2} - \cancel{Z_B^2}
$$

$$
det(Z)= Z_A \cdot Z_C + Z_A \cdot Z_B + Z_B \cdot Z_C
$$

* Calculo de **Adj(Z)**

$$
Adj(A)=
\begin{bmatrix}
Z_{22}   &    -Z_{21}  \\
-Z_{12}   &   Z_{11}     
\end{bmatrix} = 
\begin{bmatrix}
Z_{22}   &    -Z_{21}  \\
-Z_{12}   &   Z_{11}     
\end{bmatrix}
$$

Por lo que nuestra matriz de admitancias quedaria de la siguiente forma.

$$
Y = \frac{1}{det(Z)} \cdot Adj(A) =
\frac{1}{det(Z)} \cdot
\begin{bmatrix}
Z_{22}   &    -Z_{12}  \\
-Z_{21}   &   Z_{11}     
\end{bmatrix} =
\frac{1}{det(Z)}
\begin{bmatrix}
Z_B + Z_C   &    -Z_B  \\
-Z_B   &   Z_A + Z_B     
\end{bmatrix}
$$

$$
Y = 
\begin{bmatrix}
\frac{Z_B + Z_C}{Z_A \cdot Z_C + Z_A \cdot Z_B + Z_B \cdot Z_C}   &    -\frac{Z_B}{Z_A \cdot Z_C + Z_A \cdot Z_B + Z_B \cdot Z_C}  \\
-\frac{Z_B}{Z_A \cdot Z_C + Z_A \cdot Z_B + Z_B \cdot Z_C}   &   \frac{Z_A + Z_B}{Z_A \cdot Z_C + Z_A \cdot Z_B + Z_B \cdot Z_C}     
\end{bmatrix}
$$

Para finalizar esta topologia conseguiremos nuestra matriz de parametros **ABCD**, para ello tenemos que tener en cuenta su modelo de ecuacion que lo define.

$$
\begin{cases}
V_1 = I_1 \cdot Z_{11} + I_2 \cdot Z_{12} \\
V_2 = I_1 \cdot Z_{21} + I_2 \cdot Z_{22}
\end{cases}
\qquad\qquad
\begin{cases}
V_1 = V_2 \cdot A + (-I_2) \cdot B \\
I_1 = V_2 \cdot C + (-I_2) \cdot D
\end{cases}
$$

#### Cálculo del parámetro $A$

$$
A = \left. \frac{V_1}{V_2} \right|_{-I_2 = 0} \rightarrow \begin{cases} V_1 = I_1 \cdot Z_{11} \\ V_2 = I_1 \cdot Z_{21} \end{cases} \Bigg|_{-I_2 = 0} \quad \rightarrow \quad A = \frac{\cancel{I_1} \cdot Z_{11}}{\cancel{I_1} \cdot Z_{21}} = \frac{Z_{11}}{Z_{21}}
$$

#### Cálculo del parámetro $B$

$$
B = \left. \frac{V_1}{I_2} \right|_{V_2 = 0} \rightarrow 
\begin{cases} 
\overbrace{V_1 = I_1 \cdot Z_{11} + I_2 \cdot Z_{12}}^{(1)} \\ 
\underbrace{0 = I_1 \cdot Z_{21} + I_2 \cdot Z_{22}}_{(2)} 
\end{cases} \Bigg|_{V_2 = 0}
$$

De la ecuación $(2)$:

$$
0 = I_1 \cdot Z_{21} + I_2 \cdot Z_{22} \quad \rightarrow \quad I_1 \cdot Z_{21} = -I_2 \cdot Z_{22} \quad \rightarrow \quad \left[ I_1 = -I_2 \cdot \frac{Z_{22}}{Z_{21}} \right] \quad (3)
$$

Sustituyendo $(3) \rightarrow (1)$:

$$
V_1 = -I_2 \cdot \frac{Z_{22} \cdot Z_{11}}{Z_{21}} + I_2 \cdot Z_{12} = -I_2 \left( \frac{Z_{22} \cdot Z_{11} - Z_{12} \cdot Z_{21}}{Z_{21}} \right)
$$

Sabiendo que $\det(Z) = Z_{11} \cdot Z_{22} - Z_{12} \cdot Z_{21}$:

$$
\rightarrow \frac{V_1}{-I_2} = \frac{\det(Z)}{Z_{21}} \quad \rightarrow \quad B = \left. \frac{V_1}{-I_2} \right|_{V_2 = 0} = \frac{\det(Z)}{Z_{21}}
$$

#### Cálculo del parámetro $C$

$$
C = \left. \frac{I_1}{V_2} \right|_{-I_2 = 0} \rightarrow 
\begin{cases}
V_1 = I_1 \cdot Z_{11} \\
V_2 = I_1 \cdot Z_{21}
\end{cases} \Bigg|_{-I_2 = 0}
\rightarrow
C = \frac{\cancel{I_1}}{\cancel{I_1} \cdot Z_{21}} \quad \rightarrow \quad C = \frac{1}{Z_{21}}
$$

#### Cálculo del parámetro $D$

$$
D = \left. \frac{I_1}{-I_2} \right|_{V_2 = 0} \rightarrow 
\begin{cases} 
\overbrace{V_1 = I_1 \cdot Z_{11} + I_2 \cdot Z_{12}}^{(1)} \\ 
\underbrace{0 = I_1 \cdot Z_{21} + I_2 \cdot Z_{22}}_{(2)} 
\end{cases} \Bigg|_{V_2 = 0}
$$

De la ecuación $(2)$:

$$
0 = I_1 \cdot Z_{21} + I_2 \cdot Z_{22} \quad \rightarrow \quad I_1 \cdot Z_{21} = -I_2 \cdot Z_{22}
$$

Despejando la relación $\frac{I_1}{-I_2}$:

$$
I_1 \cdot Z_{21} = (- I_2) \cdot -Z_{22} \quad \rightarrow \quad \frac{I_1}{-I_2} = \frac{Z_{22}}{Z_{21}}
$$

Por lo tanto:

$$
D = \left. \frac{I_1}{-I_2} \right|_{V_2 = 0} = \frac{Z_{22}}{Z_{21}}
$$

Con los parametros calculados, nuestra matriz T queda de la siguiente forma:

$$
T= \begin{bmatrix} \frac{Z_{11}}{Z_{21}} & \frac{det(Z)}{Z_{21}} \\ \frac{1}{Z_{21}} & \frac{Z_{22}}{Z_{21}} \end{bmatrix} =\frac{1}{Z_{21}} \begin{bmatrix} Z_{11} & det(Z) \\ 1 & Z_{22} \end{bmatrix}
$$

$$
T=\frac{1}{Z_B} \begin{bmatrix} Z_A + Z_B &  Z_A \cdot Z_C + Z_A \cdot Z_B + Z_B \cdot Z_C \\ 1 & Z_C + Z_B \end{bmatrix}
$$

Entonces la **matriz T** es:

$$
T=
\begin{bmatrix}
\frac{Z_A}{Z_B} + 1 &  \frac{Z_A \cdot Z_C}{Z_B} + Z_A \cdot + Z_C \\ \frac{1}{Z_B} & \frac{Z_C}{Z_b}+1 
\end{bmatrix}
$$

### Análisis de la red tipo $\pi$

<img src="./topologia_pi.png" />

Para analizar esta red nos conviene empezar por los parámetros de admitancia ($Y$), donde conectamos una fuente de tensión en la entrada y otra en la salida, de forma tal que aplicando las definiciones de los parámetros de admitancia procedemos a calcular nuestra matriz de admitancias.

$$I_1 = V_1 \cdot Y_{11} + V_2 \cdot Y_{12}$$
$$I_2 = V_1 \cdot Y_{21} + V_2 \cdot Y_{22}$$

Donde:

$$
\begin{aligned}
Y_{11} = \left. \frac{I_1}{V_1} \right|_{V_2 = 0} &\hspace{5em} Y_{12} = \left. \frac{I_1}{V_2} \right|_{V_1 = 0} \\
\\
Y_{21} = \left. \frac{I_2}{V_1} \right|_{V_2 = 0} &\hspace{5em} Y_{22} = \left. \frac{I_2}{V_2} \right|_{V_1 = 0} \\
\end{aligned}
$$

Por lo que las componentes de la matriz quedan de la siguiente forma:

$$
\begin{aligned}
Y_{11} = \left. \frac{I_1}{V_1} \right|_{V_2 = 0} = Y_A + Y_C &\hspace{5em} Y_{12} = \left. \frac{I_1}{V_2} \right|_{V_1 = 0} = -Y_C\\
\\
Y_{22} = \left. \frac{I_2}{V_2} \right|_{V_1 = 0} = Y_B + Y_C &\hspace{5em} Y_{21} = \left. \frac{I_2}{V_1} \right|_{V_2 = 0} = -Y_C\\
\end{aligned}
$$

Quedando nuestra matriz de parámetros $Y$ de la siguiente forma:

$$
Y =
\begin{bmatrix}
Y_{11}   &    Y_{12}  \\
Y_{21}   &   Y_{22}     
\end{bmatrix} = 
\begin{bmatrix}
Y_A + Y_C   &    -Y_C  \\
-Y_C   &   Y_B + Y_C     
\end{bmatrix}
$$

Con nuestra matriz $Y$ podemos calcular nuestra matriz de impedancias $Z$, donde sabemos que $Z = Y^{-1}$:

$$
Z = \frac{1}{\det(Y)} \cdot \text{Adj}(Y)
$$

* Cálculo de **$\det(Y)$**

$$
\det(Y) = (Y_A + Y_C) \cdot (Y_B + Y_C) - (-Y_C)^2
$$

$$
\det(Y) = Y_A \cdot Y_B + Y_A \cdot Y_C + Y_B \cdot Y_C + \cancel{Y_C^2} - \cancel{Y_C^2}
$$

$$
\det(Y) = Y_A \cdot Y_B + Y_A \cdot Y_C + Y_B \cdot Y_C
$$

* Cálculo de **$\text{Adj}(Y)$**

$$
\text{Adj}(Y) =
\begin{bmatrix}
Y_{22}   &    -Y_{12}  \\
-Y_{21}   &   Y_{11}     
\end{bmatrix} = 
\begin{bmatrix}
Y_B + Y_C   &    Y_C  \\
Y_C   &   Y_A + Y_C     
\end{bmatrix}
$$

Por lo que nuestra matriz de impedancias quedaría de la siguiente forma:

$$
Z = \frac{1}{\det(Y)} \cdot \text{Adj}(Y) =
\frac{1}{\det(Y)} \cdot
\begin{bmatrix}
Y_B + Y_C   &    Y_C  \\
Y_C   &   Y_A + Y_C     
\end{bmatrix}
$$

$$
Z = 
\begin{bmatrix}
\frac{Y_B + Y_C}{Y_A \cdot Y_B + Y_A \cdot Y_C + Y_B \cdot Y_C}   &   \frac{Y_C}{Y_A \cdot Y_B + Y_A \cdot Y_C + Y_B \cdot Y_C}  \\
\frac{Y_C}{Y_A \cdot Y_B + Y_A \cdot Y_C + Y_B \cdot Y_C}   &   \frac{Y_A + Y_C}{Y_A \cdot Y_B + Y_A \cdot Y_C + Y_B \cdot Y_C}     
\end{bmatrix}
$$

Para finalizar esta topología obtendremos nuestra matriz de parámetros **ABCD** (matriz $T$), estableciendo las relaciones con las ecuaciones del modelo de admitancias $Y$:

$$
\begin{cases}
I_1 = V_1 \cdot Y_{11} + V_2 \cdot Y_{12} \\
I_2 = V_1 \cdot Y_{21} + V_2 \cdot Y_{22}
\end{cases}
\qquad\qquad
\begin{cases}
V_1 = V_2 \cdot A + (-I_2) \cdot B \\
I_1 = V_2 \cdot C + (-I_2) \cdot D
\end{cases}
$$

#### Cálculo del parámetro $A$

$$
A = \left. \frac{V_1}{V_2} \right|_{-I_2 = 0} \rightarrow 
\begin{cases} 
I_1 = V_1 \cdot Y_{11} + V_2 \cdot Y_{12} \\ 
0 = V_1 \cdot Y_{21} + V_2 \cdot Y_{22} 
\end{cases} \Bigg|_{-I_2 = 0} 
\quad \rightarrow \quad V_1 \cdot Y_{21} = -V_2 \cdot Y_{22} \quad \rightarrow \quad A = \frac{V_1}{V_2} = -\frac{Y_{22}}{Y_{21}}
$$

#### Cálculo del parámetro $B$

$$
B = \left. \frac{V_1}{-I_2} \right|_{V_2 = 0} \rightarrow 
\begin{cases} 
I_1 = V_1 \cdot Y_{11} \\ 
I_2 = V_1 \cdot Y_{21} 
\end{cases} \Bigg|_{V_2 = 0} 
\quad \rightarrow \quad -I_2 = -V_1 \cdot Y_{21} \quad \rightarrow \quad B = \frac{V_1}{-I_2} = -\frac{1}{Y_{21}}
$$

#### Cálculo del parámetro $C$

$$
C = \left. \frac{I_1}{V_2} \right|_{-I_2 = 0} \rightarrow 
\begin{cases} 
\overbrace{I_1 = V_1 \cdot Y_{11} + V_2 \cdot Y_{12}}^{(1)} \\ 
\underbrace{0 = V_1 \cdot Y_{21} + V_2 \cdot Y_{22}}_{(2)} 
\end{cases} \Bigg|_{-I_2 = 0}
$$

De la ecuación $(2)$:

$$
0 = V_1 \cdot Y_{21} + V_2 \cdot Y_{22} \quad \rightarrow \quad \left[ V_1 = -V_2 \cdot \frac{Y_{22}}{Y_{21}} \right] \quad (3)
$$

Sustituyendo $(3) \rightarrow (1)$:

$$
I_1 = \left( -V_2 \cdot \frac{Y_{22}}{Y_{21}} \right) \cdot Y_{11} + V_2 \cdot Y_{12} = -V_2 \left( \frac{Y_{11} \cdot Y_{22} - Y_{12} \cdot Y_{21}}{Y_{21}} \right)
$$

Sabiendo que $\det(Y) = Y_{11} \cdot Y_{22} - Y_{12} \cdot Y_{21}$:

$$
\rightarrow \frac{I_1}{V_2} = -\frac{\det(Y)}{Y_{21}} \quad \rightarrow \quad C = \left. \frac{I_1}{V_2} \right|_{-I_2 = 0} = -\frac{\det(Y)}{Y_{21}}
$$

#### Cálculo del parámetro $D$

$$
D = \left. \frac{I_1}{-I_2} \right|_{V_2 = 0} \rightarrow 
\begin{cases} 
\overbrace{I_1 = V_1 \cdot Y_{11}}^{(1)} \\ 
\underbrace{I_2 = V_1 \cdot Y_{21}}_{(2)} 
\end{cases} \Bigg|_{V_2 = 0}
$$

De las ecuaciones $(1)$ y $(2)$:

$$
\frac{I_1}{I_2} = \frac{\cancel{V_1} \cdot Y_{11}}{\cancel{V_1} \cdot Y_{21}} = \frac{Y_{11}}{Y_{21}}
$$

Despejando la relación $\frac{I_1}{-I_2}$:

$$
\frac{I_1}{-I_2} = -\frac{Y_{11}}{Y_{21}}
$$

Por lo tanto:

$$
D = \left. \frac{I_1}{-I_2} \right|_{V_2 = 0} = -\frac{Y_{11}}{Y_{21}}
$$

Con los parámetros calculados, nuestra matriz $T$ queda de la siguiente forma:

$$
T = \begin{bmatrix} -\frac{Y_{22}}{Y_{21}} & -\frac{1}{Y_{21}} \\ -\frac{\det(Y)}{Y_{21}} & -\frac{Y_{11}}{Y_{21}} \end{bmatrix} = -\frac{1}{Y_{21}} \begin{bmatrix} Y_{22} & 1 \\ \det(Y) & Y_{11} \end{bmatrix}
$$

Como $Y_{21} = -Y_C$:

$$
T = \frac{1}{Y_C} \begin{bmatrix} Y_B + Y_C & 1 \\ Y_A \cdot Y_B + Y_A \cdot Y_C + Y_B \cdot Y_C & Y_A + Y_C \end{bmatrix}
$$

Entonces la **matriz T** es:

$$
T = 
\begin{bmatrix}
\frac{Y_B}{Y_C} + 1 & \frac{1}{Y_C} \\ 
\frac{Y_A \cdot Y_B}{Y_C} + Y_A + Y_B & \frac{Y_A}{Y_C} + 1 
\end{bmatrix}
$$

### Análisis de la red tipo Lattice

<img src="./topologia_lattice.png" />

Para analizar esta red simétrica, comenzamos por los parámetros de impedancia ($Z$). Conectamos una fuente de corriente en la entrada ($I_1$) y otra en la salida ($I_2$), de forma tal que aplicando las definiciones de los parámetros de impedancia procedemos a calcular nuestra matriz de impedancia:

$$V_1 = I_1 \cdot Z_{11} + I_2 \cdot Z_{12}$$
$$V_2 = I_1 \cdot Z_{21} + I_2 \cdot Z_{22}$$

Donde:

$$
\begin{aligned}
Z_{11} = \left. \frac{V_1}{I_1} \right|_{I_2 = 0} &\hspace{5em} Z_{12} = \left. \frac{V_1}{I_2} \right|_{I_1 = 0} \\
\\
Z_{21} = \left. \frac{V_2}{I_1} \right|_{I_2 = 0} &\hspace{5em} Z_{22} = \left. \frac{V_2}{I_2} \right|_{I_1 = 0} \\
\end{aligned}
$$

Por simetría del puente, cuando $I_2 = 0$, la corriente $I_1$ se divide por igual entre las dos ramas paralelas compuestas por la serie $(Z_A + Z_B)$. Calculando $V_1$ y $V_2$ respecto a la referencia común:

$$
\begin{aligned}
Z_{11} = \left. \frac{V_1}{I_1} \right|_{I_2 = 0} = \frac{Z_A + Z_B}{2} &\hspace{5em} Z_{12} = \left. \frac{V_1}{I_2} \right|_{I_1 = 0} = \frac{Z_B - Z_A}{2} \\
\\
Z_{22} = \left. \frac{V_2}{I_2} \right|_{I_1 = 0} = \frac{Z_A + Z_B}{2} &\hspace{5em} Z_{21} = \left. \frac{V_2}{I_1} \right|_{I_2 = 0} = \frac{Z_B - Z_A}{2} \\
\end{aligned}
$$

Quedando nuestra matriz de parámetros $Z$ de la siguiente forma:

$$
Z =
\begin{bmatrix}
Z_{11}   &    Z_{12}  \\
Z_{21}   &   Z_{22}     
\end{bmatrix} = 
\begin{bmatrix}
\frac{Z_B + Z_A}{2}   &   \frac{Z_B - Z_A}{2}  \\
\frac{Z_B - Z_A}{2}   &   \frac{Z_B + Z_A}{2}     
\end{bmatrix}
$$

Con nuestra matriz $Z$ podemos calcular nuestra matriz de admitancias $Y$, donde sabemos que $Y = Z^{-1}$:

$$
Y = \frac{1}{\det(Z)} \cdot \text{Adj}(Z)
$$

* Cálculo de **$\det(Z)$**

$$
\det(Z) = \left(\frac{Z_B + Z_A}{2}\right)^2 - \left(\frac{Z_B - Z_A}{2}\right)^2
$$

$$
\det(Z) = \frac{Z_B^2 + 2 Z_A Z_B + Z_A^2}{4} - \frac{Z_B^2 - 2 Z_A Z_B + Z_A^2}{4} = \frac{4 Z_A Z_B}{4} = Z_A \cdot Z_B
$$

$$
\det(Z) = Z_A \cdot Z_B
$$

* Cálculo de **$\text{Adj}(Z)$**

$$
\text{Adj}(Z) =
\begin{bmatrix}
Z_{22}   &    -Z_{12}  \\
-Z_{21}   &   Z_{11}     
\end{bmatrix} = 
\begin{bmatrix}
\frac{Z_B + Z_A}{2}   &   \frac{Z_A - Z_B}{2}  \\
\frac{Z_A - Z_B}{2}   &   \frac{Z_B + Z_A}{2}     
\end{bmatrix}
$$

Por lo que nuestra matriz de admitancias quedaría de la siguiente forma:

$$
Y = \frac{1}{\det(Z)} \cdot \text{Adj}(Z) =
\frac{1}{Z_A \cdot Z_B} \cdot
\begin{bmatrix}
\frac{Z_B + Z_A}{2}   &   \frac{Z_A - Z_B}{2}  \\
\frac{Z_A - Z_B}{2}   &   \frac{Z_B + Z_A}{2}     
\end{bmatrix}
$$

$$
Y = 
\begin{bmatrix}
\frac{1}{2 Z_A} + \frac{1}{2 Z_B}   &   \frac{1}{2 Z_B} - \frac{1}{2 Z_A}  \\
\frac{1}{2 Z_B} - \frac{1}{2 Z_A}   &   \frac{1}{2 Z_A} + \frac{1}{2 Z_B}     
\end{bmatrix}
$$

Donde podemos observar que si cambiamos las impedancias por admitancias la matriz nos queda igual que la de parametros Z.

$$
Y=
\begin{bmatrix}
\frac{Y_A+Y_B}{2}     &   \frac{Y_B-Y_A}{2} \\
\frac{Y_B-Y_A}{2}  &   \frac{Y_A+Y_B}{2}       
\end{bmatrix}
$$

Para finalizar esta topología obtendremos nuestra matriz de parámetros **ABCD** (matriz $T$), usando las ecuaciones calculadas anteriormente (Red tipo T):

$$
T = \begin{bmatrix} \frac{Z_{11}}{Z_{21}} & \frac{\det(Z)}{Z_{21}} \\ \frac{1}{Z_{21}} & \frac{Z_{22}}{Z_{21}} \end{bmatrix} = \frac{1}{Z_{21}} \begin{bmatrix} Z_{11} & \det(Z) \\ 1 & Z_{22} \end{bmatrix}
$$

Sustituyendo las componentes de $Z$ obtenidas para la red Lattice:

$$
T = \frac{1}{\frac{Z_B - Z_A}{2}} \begin{bmatrix} \frac{Z_B + Z_A}{2} & Z_A \cdot Z_B \\ 1 & \frac{Z_B + Z_A}{2} \end{bmatrix}
$$

Entonces la **matriz T** es:

$$
